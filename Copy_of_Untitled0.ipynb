```python
# @title Default title text
import statsmodels.api as sm
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
from scipy import stats

df = pd.read_csv('group_05.csv')


df_4=df.replace(to_replace={'Male':0, 'Female':1})
df_4["gndr"] = pd.to_numeric(df_4["gndr"])
model_4=df_4[['agea', 'gndr', 'eduyrs']]
model_4=sm.add_constant(model_4)
result4=sm.OLS(df_4['stflife'], model_4, missing='drop').fit()
print(result4.summary())

df_new=df.replace(to_replace={'Male':0, 'Female':1})
df_new["gndr"] = pd.to_numeric(df_new["gndr"])
model_1=df_new[['agea', 'gndr', 'hinctnta']]
model_1=sm.add_constant(model_1)
result1=sm.OLS(df_new['stflife'], model_1, missing='drop').fit()
print(result1.summary())


edu_squared=pd.Series(map(lambda x: x**2, df['eduyrs']))
df_for_model_2=df.copy()
df_for_model_2.insert(5, "eduyrs_2", edu_squared)
model_2=df_for_model_2[['eduyrs', 'eduyrs_2','agea']]
model_2=sm.add_constant(model_2)
result2=sm.OLS(df_new['stflife'], model_2, missing='drop').fit()
print(result2.summary())

inter=[]
for i in range(len(df['ppltrst'])):
        inter.append(df['ppltrst'][i]*df['sclmeet'][i])
interaction_term=pd.Series(inter)
df_for_model_3=df.copy()
df_for_model_3.insert(5, "interaction_term", interaction_term)
model_3=df_for_model_3[['agea','ppltrst','sclmeet', 'interaction_term']]
model_3=sm.add_constant(model_3)
result3=sm.OLS(df_new['stflife'], model_3, missing='drop').fit()
print(result3.summary())







```
